```python
import matplotlib.pyplot as plt
import numpy as np

methods = ["FaRM", "FiLM", "Concatenation", "Unconditioned"]
colors = ["red", "purple", "green", "blue"]

gpu_memory_gb = [5679000 / 1e6, 7793000 / 1e6, 5397000 / 1e6, 5681000 / 1e6]
step_time = [60 * 1.20514 / 314, 60 * 1.68088 / 314, 60 * 1.18692 / 314, 60 * 1.20514 / 314]

n = len(methods)
width = 0.18

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(6, 4))

for i in range(n):
    offset = (i - (n - 1) / 2) * width
    ax1.bar(offset, gpu_memory_gb[i], width=width, color=colors[i], label=methods[i])
    ax2.bar(offset, step_time[i], width=width, color=colors[i], label=methods[i])

for ax, ylabel, xlabel in [(ax1, "GB", "GPU Memory"), (ax2, "Seconds", "Step Time")]:
    ax.set_xticks([])
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.set_xlim(-0.5, 0.5)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

ax1.legend(fontsize=8,loc="upper right")
ax2.legend(fontsize=8,loc="upper right")
fig.suptitle("Training resources, FaRM vs FiLM vs Concatenation", fontsize=12)
plt.tight_layout()
plt.savefig("pdfs/supplement__conditioning_compute_comparison.pdf", bbox_inches="tight")
plt.show()
```
